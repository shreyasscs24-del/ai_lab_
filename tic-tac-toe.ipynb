```python
# Tic-Tac-Toe using Game Tree and Minimax

board = [" "] * 9


# Display the board
def display_board():
    print()
    print(" {} | {} | {}".format(board[0], board[1], board[2]))
    print("---+---+---")
    print(" {} | {} | {}".format(board[3], board[4], board[5]))
    print("---+---+---")
    print(" {} | {} | {}".format(board[6], board[7], board[8]))
    print()


# Check whether a player has won
def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


# Check whether board is full
def is_full():
    return " " not in board


# Minimax algorithm
def minimax(is_maximizing):

    # Terminal states
    if check_winner("X"):
        return 1

    if check_winner("O"):
        return -1

    if is_full():
        return 0

    # X tries to maximize the score
    if is_maximizing:
        best_score = -100

        for i in range(9):
            if board[i] == " ":
                board[i] = "X"

                score = minimax(False)

                board[i] = " "

                best_score = max(best_score, score)

        return best_score

    # O tries to minimize the score
    else:
        best_score = 100

        for i in range(9):
            if board[i] == " ":
                board[i] = "O"

                score = minimax(True)

                board[i] = " "

                best_score = min(best_score, score)

        return best_score


# Find the best move for computer
def find_best_move():
    best_score = -100
    best_move = -1

    for i in range(9):
        if board[i] == " ":
            board[i] = "X"

            score = minimax(False)

            board[i] = " "

            if score > best_score:
                best_score = score
                best_move = i

    return best_move


# Main program
print("TIC-TAC-TOE")
print("Computer = X")
print("Player   = O")

print("\nPositions:")
print(" 1 | 2 | 3")
print("---+---+---")
print(" 4 | 5 | 6")
print("---+---+---")
print(" 7 | 8 | 9")

while True:

    # Computer's turn
    computer_move = find_best_move()
    board[computer_move] = "X"

    print("\nComputer played:", computer_move + 1)
    display_board()

    if check_winner("X"):
        print("Computer (X) wins!")
        break

    if is_full():
        print("Game is a Draw!")
        break

    # Player's turn
    while True:
        try:
            player_move = int(input("Enter your position (1-9): "))

            if player_move < 1 or player_move > 9:
                print("Enter a number between 1 and 9.")
            elif board[player_move - 1] != " ":
                print("Position already occupied. Try again.")
            else:
                break

        except ValueError:
            print("Please enter a valid number.")

    board[player_move - 1] = "O"
    display_board()

    if check_winner("O"):
        print("Player (O) wins!")
        break

    if is_full():
        print("Game is a Draw!")
        break
```
