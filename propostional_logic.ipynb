```python
from itertools import product

def implies(a, b):
    return not a or b


def knowledge_base(P, Q, R):
    return (implies(Q, P) and
            implies(P, not Q) and
            (Q or R))


def truth_table():
    entail_R = True
    entail_RP = True
    entail_QR = True

    print("P Q R | KB | R | R->P | Q->R")
    print("--------------------------------")

    for P, Q, R in product([False, True], repeat=3):
        kb = knowledge_base(P, Q, R)

        q1 = R
        q2 = implies(R, P)
        q3 = implies(Q, R)

        print(int(P), int(Q), int(R), "|",
              int(kb), "|", int(q1), "|",
              int(q2), "    |", int(q3))

        if kb:
            if not q1:
                entail_R = False
            if not q2:
                entail_RP = False
            if not q3:
                entail_QR = False

    print("\nEntailment Results:")
    print("KB entails R:", entail_R)
    print("KB entails R -> P:", entail_RP)
    print("KB entails Q -> R:", entail_QR)


truth_table()
```
