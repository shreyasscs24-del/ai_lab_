```python
def cost(state):
    count = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                count += 1

    return count


def display(state):
    n = len(state)
    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
    print()


def hill_climbing(state):
    step = 0

    while True:
        print("Step:", step)
        print("Current State:", state)
        print("Cost:", cost(state))
        display(state)

        if cost(state) == 0:
            print("Goal Reached!")
            break

        best_state = state[:]
        best_cost = cost(state)

        for i in range(len(state)):
            for j in range(i + 1, len(state)):
                neighbor = state[:]
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                h = cost(neighbor)
                print("Neighbor:", neighbor, "Cost:", h)

                if h < best_cost:
                    best_cost = h
                    best_state = neighbor

        if best_cost >= cost(state):
            print("No better neighbor found.")
            break

        state = best_state
        step += 1
        print()


initial = [3, 1, 2, 0]
hill_climbing(initial)
```
