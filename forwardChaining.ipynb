```python

# Forward Chaining using First Order Logic

facts = []

# Get facts from user
n = int(input("Enter number of facts: "))

for i in range(n):
    print("Enter fact", i + 1)
    x, y = input("Enter Parent (person1 person2): ").split()
    facts.append(("Parent", x, y))

# Get query from user
print("\nEnter Query:")
x, y = input("Enter Grandparent (person1 person2): ").split()
query = ("Grandparent", x, y)

print("\nInitial Facts:")
for f in facts:
    print(f[0] + "(" + f[1] + ", " + f[2] + ")")

print("\nApplying Forward Chaining...\n")

new_facts = []

# Apply rule:
# Parent(X,Y) AND Parent(Y,Z) -> Grandparent(X,Z)

for f1 in facts:
    for f2 in facts:

        if f1[2] == f2[1]:

            new_fact = ("Grandparent", f1[1], f2[2])

            if new_fact not in facts and new_fact not in new_facts:
                new_facts.append(new_fact)

                print("New Fact Generated:")
                print("Grandparent(" + f1[1] + ", " + f2[2] + ")")

facts.extend(new_facts)

print("\nChecking Query:")

if query in facts:
    print("Query Proved Successfully!")
    print("Result: TRUE")
else:
    print("Query Not Proved!")
    print("Result: FALSE")
```
