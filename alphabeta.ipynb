```python

def get_leaves(node):
    if isinstance(node, int):
        return [node]

    leaves = []
    for child in node[1]:
        leaves.extend(get_leaves(child))
    return leaves


def alpha_beta(node, alpha, beta, maximizing):
    if isinstance(node, int):
        return node

    children = node[1]

    if maximizing:
        value = float('-inf')

        for i, child in enumerate(children):
            value = max(value, alpha_beta(child, alpha, beta, False))
            alpha = max(alpha, value)

            if alpha >= beta:
                for remaining in children[i + 1:]:
                    pruned.extend(get_leaves(remaining))
                break

        return value

    else:
        value = float('inf')

        for i, child in enumerate(children):
            value = min(value, alpha_beta(child, alpha, beta, True))
            beta = min(beta, value)

            if alpha >= beta:
                for remaining in children[i + 1:]:
                    pruned.extend(get_leaves(remaining))
                break

        return value


# Game tree from the PPT
tree = ("MAX", [
    ("MIN", [
        ("MAX", [10, 9]),
        ("MAX", [14, 18])
    ]),
    ("MIN", [
        ("MAX", [5, 4]),
        ("MAX", [50, 3])
    ])
])

pruned = []

result = alpha_beta(tree, float('-inf'), float('inf'), True)

print("Optimal value of root node:", result)
print("Pruned leaf nodes:", pruned)
```
