```python

def unify(x, y, subs):

    # Apply existing substitutions
    while x in subs:
        x = subs[x]

    while y in subs:
        y = subs[y]

    if x == y:
        return True

    # Check whether x is a variable
    if x.islower():
        subs[x] = y
        return True

    # Check whether y is a variable
    if y.islower():
        subs[y] = x
        return True

    return False


# Input
expr1 = input("Enter first expression: ")
expr2 = input("Enter second expression: ")

def extract(expr):
    name, args = expr.split("(")
    args = args.replace(")", "").split(",")
    return name.strip(), [a.strip() for a in args]


p1, a1 = extract(expr1)
p2, a2 = extract(expr2)

subs = {}

# Perform unification
if p1 != p2 or len(a1) != len(a2):
    print("\nUnification Failed")

else:
    success = True

    for x, y in zip(a1, a2):
        if not unify(x, y, subs):
            success = False
            break

    if success:
        print("\nUnification Successful")
        print("Substitution:")

        for x, y in subs.items():
            print(x, "=", y)

        result = [subs.get(x, x) for x in a1]

        print("\nUnified Expression:")
        print(p1 + "(" + ", ".join(result) + ")")

    else:
        print("\nUnification Failed")
```
